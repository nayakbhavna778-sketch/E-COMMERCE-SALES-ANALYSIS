# 🛒 E-commerce Sales Analytics: Data Cleaning (Step 1)

## Objective
Clean the raw e-commerce sales dataset so it is ready for analysis and dashboarding in the next steps.

## Dataset Overview
- **File:** `ecommerce_sales_analytics_5000.csv`
- **Size:** 5,000 orders × 12 columns

| Column | Description |
|--------|-------------|
| `order_id` | Unique ID of each order |
| `order_date` | Date the order was placed |
| `customer_id` | Unique ID of the customer |
| `product_category` | Category of the product (Beauty, Clothing, Electronics, Home, ...) |
| `region` | Region of the order (North, South, East, West) |
| `quantity` | Number of units ordered |
| `unit_price` | Price per unit |
| `discount` | Discount applied (as a fraction, e.g. 0.28 = 28%) |
| `payment_method` | Mode of payment (Wallet, COD, ...) |
| `delivery_days` | Days taken to deliver the order |
| `customer_rating` | Customer rating for the order |
| `revenue` | Total revenue from the order |

## Cleaning Steps
1. Import libraries and load the data
2. Inspect the data (`head`, `tail`, `info`)
3. Convert `order_date` to datetime
4. Check for missing values
5. Check for duplicate rows
6. Save the cleaned dataset
7. Verify and download the output

## 1. Import Libraries & Load Data
We use **pandas** for data handling. The raw CSV is loaded into a DataFrame called `df`.

```python
import pandas as pd
df=pd.read_csv("ecommerce_sales_analytics_5000.csv")
print(df)
```

## 2. Initial Data Inspection
Before cleaning, we get a first look at the data's structure and values.

### 2.1 First 5 Rows
`df.head()` shows the top rows so we can see the column names and the kind of values stored in each.

```python
df.head()
```

### 2.2 Last 5 Rows
`df.tail()` shows the bottom rows to confirm the data was loaded completely and to check how the dates and IDs progress towards the end.

> **Note:** The last rows show order dates going up to **2035**, which is far in the future compared to the start of the data (2022). Worth keeping in mind during analysis.

```python
df.tail()
```

### 2.3 Data Types & Non-Null Counts
`df.info()` gives the number of rows, the data type of every column and the count of non-null values.

**Observations:**
- The dataset has **5,000 rows and 12 columns**.
- All columns show 5,000 non-null values, so there are no missing entries.
- `order_date` is stored as `object` (plain text), so it needs to be converted to a proper date type.

```python
df.info()
```

## 3. Fix Data Types
### Convert `order_date` to Datetime
The date column is currently text, which prevents date-based operations. Converting it with `pd.to_datetime()` lets us extract the year, month and day, sort by time and build trends later.

```python
df['order_date']=pd.to_datetime(df['order_date'])
```

## 4. Check for Missing Values
`df.isnull().sum()` counts the null values in each column.

**Result:** Every column has **0 missing values**, so no imputation or row removal is needed.

```python
df.isnull().sum()
```

## 5. Check for Duplicate Records
`df.duplicated().sum()` counts rows that are exact copies of another row.

**Result:** **0 duplicates** found, so every order record is unique.

```python
df.duplicated().sum()
```

## 6. Save the Cleaned Dataset
The cleaned DataFrame is exported to `ecommerce_sales_analytics_5000_cleaned.csv`. We use `index=False` so the row index is not saved as an extra column.

```python
df.to_csv('ecommerce_sales_analytics_5000_cleaned.csv', index=False)
```
